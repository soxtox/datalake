# business transformation and modelling



```python
query = """
SELECT 
Row_number() over (partition by ci.firstname,lastname order by ci.customer_id desc) as customer_rnk,
ci.customer_id,
ci.customer_key,
ci.firstname,
ci.lastname,
ci.marital_status,
ci.gender,
ci.create_date
FROM silver.customer_info ci
"""


df = spark.sql(query) 
df.display()
```

# write to gold layer

```python
df.write.mode("overwrite").saveAsTable("Gold.customer_info")
```
