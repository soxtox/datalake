###### So we need to create the dictionary file and iterate over the file to create the tables

# ingest file into Bronze tables
volume name shall not contain underscore

```python
from bronze_config import INGESTION_CONFIG
```

```python
%sql
USE CATALOG workspace;
use schema bronze
```

```python

for item in INGESTION_CONFIG:
  print(f"ingesting {item['source']} -> bronze.{item['table']}")
  df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(item["path"])
    )
  ( 
   df.write.mode("overwrite").format("delta").saveAsTable(f"bronze.{item['table']}")
   )
```
